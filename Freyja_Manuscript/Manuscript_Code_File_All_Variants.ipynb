```python
from datetime import datetime, timedelta
import pandas as pd

start_date = datetime(2021, 2, 7)

# Define the end date (Feb 27, 2022)
end_date = datetime(2022, 2, 27)

# Initialize the current date as the start date
current_date = start_date

# Define a timedelta of 7 days to increment by one week
one_week = timedelta(days=7)

# Create an empty list to store the data
data = []

# Loop to generate epiweeks and week numbers
while current_date <= end_date:
    # Calculate the epiweek and week number
    epiweek = current_date.strftime("%U")
    week_number = current_date.strftime("%W")
    date_str = current_date.strftime("%Y-%m-%d")

    # Append the epiweek, week number, and date to the list
    data.append([epiweek, week_number, date_str])

    # Increment the current date by one week
    current_date += one_week

# Create a DataFrame from the list
epiweeks_df = pd.DataFrame(data, columns=['Epiweek Number', 'Week Number', 'Date'])

epiweeks_df
```

```python
from scipy.stats import shapiro
import pandas as pd

# Original column names from dataset
original_cli_columns = ['alpha_percentage', 'beta_percentage', 'gamma_percentage', 'delta_percentage', 'epsilon_percentage', 'omicron_percentage']
original_ww_columns = ['Alpha%', 'Beta%', 'Gamma%', 'Delta%', 'Epsilon%', 'Omicron%']

# Descriptive column names for output
cli_columns = ['Alpha_Clinical', 'Beta_Clinical', 'Gamma_Clinical', 'Delta_Clinical', 'Epsilon_Clinical', 'Omicron_Clinical']
ww_columns = ['Alpha_Wastewater', 'Beta_Wastewater', 'Gamma_Wastewater', 'Delta_Wastewater', 'Epsilon_Wastewater', 'Omicron_Wastewater']

# Initialize lists to hold results for each type
cli_results = {'Variants': [], 'Shapiro Statistic': [], 'P-value': [], 'Normality Conclusion': []}
ww_results = {'Variants': [], 'Shapiro Statistic': [], 'P-value': [], 'Normality Conclusion': []}

# Assuming 'combined_study_cleandf' is your DataFrame
for original_cli, cli_col, original_ww, ww_col in zip(original_cli_columns, cli_columns, original_ww_columns, ww_columns):
    # Clinical data testing
    stat_cli, p_cli = shapiro(combined_study_cleandf[original_cli].dropna())
    cli_results['Variants'].append(cli_col)
    cli_results['Shapiro Statistic'].append(round(stat_cli, 3))
    cli_results['P-value'].append(p_cli)
    cli_conclusion = "Not Normally Distributed (Reject H0)" if p_cli < 0.05 else "Normally Distributed (Fail to reject H0)"
    cli_results['Normality Conclusion'].append(cli_conclusion)

    # Wastewater data testing
    stat_ww, p_ww = shapiro(combined_study_cleandf[original_ww].dropna())
    ww_results['Variants'].append(ww_col)
    ww_results['Shapiro Statistic'].append(round(stat_ww, 3))
    ww_results['P-value'].append(p_ww)
    ww_conclusion = "Not Normally Distributed (Reject H0)" if p_ww < 0.05 else "Normally Distributed (Fail to reject H0)"
    ww_results['Normality Conclusion'].append(ww_conclusion)

# Creating the DataFrames from the results
cli_df = pd.DataFrame(cli_results)
ww_df = pd.DataFrame(ww_results)

# Display the DataFrames
print("Clinical Variants Shapiro-Wilk Test Results:")

print("\nWastewater Variants Shapiro-Wilk Test Results:")
ww_df

```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from scipy.stats import spearmanr, pearsonr
from matplotlib.lines import Line2D
from matplotlib import font_manager

# ----------------------------------------------------
# FONT SETTINGS FOR PLOS ONE
# ----------------------------------------------------
available_fonts = [f.name for f in font_manager.fontManager.ttflist]

if 'Arial' in available_fonts:
    chosen_font = 'Arial'
elif 'Times New Roman' in available_fonts:
    chosen_font = 'Times New Roman'
else:
    chosen_font = 'Times'

plt.rcParams.update({
    'font.family': chosen_font,
    'font.size': 9,
    'axes.labelsize': 9,
    'axes.titlesize': 9,
    'xtick.labelsize': 9,
    'ytick.labelsize': 9,
    'legend.fontsize': 8,
    'mathtext.fontset': 'stix'
})

# ----------------------------------------------------
# DEFINE VARIANTS, COLUMNS, AND COLORS
# ----------------------------------------------------
cli_columns = ['alpha_percentage', 'epsilon_percentage', 'beta_percentage', 'gamma_percentage', 'delta_percentage', 'omicron_percentage']
ww_columns = ['Alpha%', 'Epsilon%', 'Beta%', 'Gamma%', 'Delta%', 'Omicron%']
variant_names = ['Alpha', 'Epsilon', 'Beta', 'Gamma', 'Delta', 'Omicron']

color_map = {
    'Alpha': 'orange',
    'Epsilon': 'aqua',
    'Beta': 'blue',
    'Gamma': 'purple',
    'Delta': 'green',
    'Omicron': 'brown'
}

# ----------------------------------------------------
# PREP DATA
# ----------------------------------------------------
combined_study_df['date'] = pd.to_datetime(combined_study_df['date'])
combined_study_df['Epiweek Number'] = combined_study_df['Epiweek Number'].astype(int)

# ----------------------------------------------------
# CREATE PANEL FIGURE
# ----------------------------------------------------
fig, axes = plt.subplots(
    2, 1,
    figsize=(7.5, 8),
    dpi=300,
    gridspec_kw={'height_ratios': [1.00, 1.20]}
)

ax1 = axes[0]
ax2 = axes[1]

# ====================================================
# FIGURE 3A: VARIANT DISTRIBUTION OVER TIME
# ====================================================
for cli_col, ww_col, name in zip(cli_columns, ww_columns, variant_names):

    color = color_map[name]

    ax1.plot(
        combined_study_df['date'],
        combined_study_df[cli_col],
        linestyle='--',
        linewidth=1.5,
        color=color
    )

    ax1.plot(
        combined_study_df['date'],
        combined_study_df[ww_col],
        linestyle='-',
        linewidth=1.5,
        color=color
    )

epiweek_filter = (combined_study_df['Epiweek Number'] - 6) % 4 == 0
filtered_dates = combined_study_df.loc[epiweek_filter, 'date']
filtered_epiweeks = combined_study_df.loc[epiweek_filter, 'Epiweek Number']

ax1.set_xticks(filtered_dates)
ax1.set_xticklabels(filtered_epiweeks, rotation=45, ha='right')

ax1.set_xlabel('Epiweek number', fontweight='bold')
ax1.set_ylabel('SARS-CoV-2 variant detection %', fontweight='bold')
ax1.set_title(
    'Distribution of SARS-CoV-2 variants in clinical and wastewater samples over time',
    fontweight='bold',
    pad=6
)

ax1.text(
    -0.08, 1.02, 'A',
    transform=ax1.transAxes,
    fontsize=12,
    fontweight='bold',
    va='top',
    ha='left'
)

variant_handles = [
    Line2D([0], [0], color=color_map[name], lw=2, label=name)
    for name in variant_names
]

legend1 = ax1.legend(
    handles=variant_handles,
    title='Variant',
    title_fontsize=8,
    loc='upper left',
    bbox_to_anchor=(1.02, 1.00),
    borderaxespad=0.0,
    frameon=True,
    fancybox=False,
    edgecolor='black'
)

ax1.add_artist(legend1)

source_handles = [
    Line2D([0], [0], color='black', lw=1.5, linestyle='--', label='Clinical'),
    Line2D([0], [0], color='black', lw=1.5, linestyle='-', label='Wastewater')
]

ax1.legend(
    handles=source_handles,
    title='Source',
    title_fontsize=8,
    loc='upper right',
    bbox_to_anchor=(1.23, 0.55),
    borderaxespad=0.0,
    frameon=True,
    fancybox=False,
    edgecolor='black'
)

# ====================================================
# FIGURE 3B: CLINICAL VS WASTEWATER CORRELATION
# ====================================================
results_list = []

all_cli_data = pd.Series(dtype=float)
all_ww_data = pd.Series(dtype=float)

for cli_col, ww_col, name in zip(cli_columns, ww_columns, variant_names):

    valid_df = combined_study_cleandf[[cli_col, ww_col]].dropna()
    X = valid_df[cli_col]
    y = valid_df[ww_col]

    color = color_map[name]

    ax2.scatter(
        X,
        y,
        color=color,
        s=28,
        alpha=0.85,
        label=name
    )

    all_cli_data = pd.concat([all_cli_data, X])
    all_ww_data = pd.concat([all_ww_data, y])

    model_variant = LinearRegression().fit(
        X.values.reshape(-1, 1),
        y.values.reshape(-1, 1)
    )
    y_pred_variant = model_variant.predict(X.values.reshape(-1, 1))

    pearson_corr, pearson_p = pearsonr(X, y)
    spearman_corr, spearman_p = spearmanr(X, y)
    r_squared = r2_score(y, y_pred_variant)

    results_list.append({
        'Variant': name,
        'r_s_Freyja': round(spearman_corr, 3),
        'p-value_Freyja': spearman_p,
        'R-squared_Freyja': round(r_squared, 3)
    })

overall_model = LinearRegression().fit(
    all_cli_data.values.reshape(-1, 1),
    all_ww_data.values.reshape(-1, 1)
)

overall_y_pred = overall_model.predict(
    all_cli_data.values.reshape(-1, 1)
).flatten()

overall_r_squared = r2_score(all_ww_data, overall_y_pred)
overall_pearson, overall_pearson_p = pearsonr(all_cli_data, all_ww_data)
overall_spearman, overall_spearman_p = spearmanr(all_cli_data, all_ww_data)

results_list.append({
    'Variant': 'All',
    'r_s_Freyja': round(overall_spearman, 3),
    'p-value_Freyja': overall_spearman_p,
    'R-squared_Freyja': round(overall_r_squared, 3)
})

sort_idx = np.argsort(all_cli_data.values)
x_sorted = all_cli_data.values[sort_idx]
y_sorted = overall_y_pred[sort_idx]

ax2.plot(
    x_sorted,
    y_sorted,
    color='black',
    linewidth=1.5,
    label=f'Overall fit\nR² = {overall_r_squared:.3f}, r$_s$ = {overall_spearman:.3f}'
)

ax2.set_xlabel(
    'Relative abundances of SARS-CoV-2 variants in clinical samples',
    fontweight='bold'
)

ax2.set_ylabel(
    'Relative abundances of SARS-CoV-2 variants\nin wastewater samples',
    fontweight='bold'
)

ax2.set_xlim(-0.05, 1.05)
ax2.set_ylim(-0.05, 1.05)

ax2.text(
    -0.08, 1.04, 'B',
    transform=ax2.transAxes,
    fontsize=12,
    fontweight='bold',
    va='top',
    ha='left'
)

ax2.legend(
    loc='upper left',
    bbox_to_anchor=(1.02,1.00),
    borderaxespad=0.0,
    frameon=True,
    fancybox=False,
    edgecolor='black'
)

# ----------------------------------------------------
# MAKE ALL TICKS, TITLES, LABELS BOLD
# ----------------------------------------------------
for ax in [ax1, ax2]:

    ax.tick_params(axis='both', which='major', labelsize=9, width=1, length=4)

    for tick in ax.get_xticklabels():
        tick.set_fontname(chosen_font)
        tick.set_fontweight('bold')

    for tick in ax.get_yticklabels():
        tick.set_fontname(chosen_font)
        tick.set_fontweight('bold')

    ax.xaxis.label.set_fontname(chosen_font)
    ax.xaxis.label.set_fontweight('bold')

    ax.yaxis.label.set_fontname(chosen_font)
    ax.yaxis.label.set_fontweight('bold')

    ax.title.set_fontname(chosen_font)
    ax.title.set_fontweight('bold')

    for spine in ax.spines.values():
        spine.set_linewidth(1)

# ----------------------------------------------------
# FINAL LAYOUT
# ----------------------------------------------------
plt.subplots_adjust(
    left=0.17,
    right=0.78,
    top=0.96,
    bottom=0.08,
    hspace=0.22
)

plt.savefig('/content/drive/MyDrive/Freyja_Manuscript/Master_File/Output/Fig3.png', dpi=300)
plt.savefig('/content/drive/MyDrive/Freyja_Manuscript/Master_File/Output/Fig3.tiff', dpi=300)

plt.show()

table2_freyja_df = pd.DataFrame(results_list)

# Move All row to the top
table2_freyja_df = pd.concat([
    table2_freyja_df[table2_freyja_df['Variant'] == 'All'],
    table2_freyja_df[table2_freyja_df['Variant'] != 'All']
], ignore_index=True)

def format_p_value(p):
    if pd.isna(p):
        return ""
    if p == 0:
        return "<1 × 10^-300"
    exponent = int(np.floor(np.log10(abs(p))))
    coefficient = p / (10 ** exponent)
    return f"{coefficient:.2f} × 10^{exponent}"

table2_freyja_display_df = table2_freyja_df.copy()

table2_freyja_display_df['p-value_Freyja'] = table2_freyja_display_df['p-value_Freyja'].apply(format_p_value)

table2_freyja_display_df
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import pandas as pd
import numpy as np
from scipy.stats import spearmanr

# ----------------------------------------------------
# DEFINE EXACT COLUMNS IN MATCHING ORDER
#
# ----------------------------------------------------
cli_columns = ['alpha_percentage', 'epsilon_percentage', 'beta_percentage', 'gamma_percentage', 'delta_percentage', 'omicron_percentage']
ww_columns = ['Alpha_MLST', 'Epsilon_MLST', 'Beta_MLST', 'Gamma_MLST', 'Delta_MLST', 'Omicron_MLST']
variant_names = ['Alpha', 'Epsilon', 'Beta', 'Gamma', 'Delta', 'Omicron']

# ----------------------------------------------------
# CHECK COLUMN PAIRS
# ----------------------------------------------------
for cli_col, ww_col, name in zip(cli_columns, ww_columns, variant_names):
    print(f"{name}: {cli_col}  vs  {ww_col}")

# ----------------------------------------------------
# CREATE MLST VS CLINICAL TABLE
# ----------------------------------------------------
mlst_results_list = []

all_cli_data = pd.Series(dtype=float)
all_mlst_data = pd.Series(dtype=float)

for cli_col, ww_col, name in zip(cli_columns, ww_columns, variant_names):

    valid_df = combined_study_df[[cli_col, ww_col]].dropna()

    X = valid_df[cli_col]
    y = valid_df[ww_col]

    # Aggregate data for overall correlation
    all_cli_data = pd.concat([all_cli_data, X], ignore_index=True)
    all_mlst_data = pd.concat([all_mlst_data, y], ignore_index=True)

    # Linear regression for R-squared
    model = LinearRegression().fit(
        X.values.reshape(-1, 1),
        y.values.reshape(-1, 1)
    )

    y_pred = model.predict(X.values.reshape(-1, 1))

    # Spearman correlation and p-value
    spearman_corr, spearman_p = spearmanr(X, y)

    # R-squared from linear regression
    r_squared = r2_score(y, y_pred)

    mlst_results_list.append({
        'Variant': name,
        'r_s_MLST': round(spearman_corr, 3),
        'p-value_MLST': spearman_p,
        'R-squared_MLST': round(r_squared, 3)
    })

# ----------------------------------------------------
# OVERALL MLST VS CLINICAL
# ----------------------------------------------------
overall_model = LinearRegression().fit(
    all_cli_data.values.reshape(-1, 1),
    all_mlst_data.values.reshape(-1, 1)
)

overall_y_pred = overall_model.predict(
    all_cli_data.values.reshape(-1, 1)
).flatten()

overall_r_squared = r2_score(all_mlst_data, overall_y_pred)
overall_spearman, overall_spearman_p = spearmanr(all_cli_data, all_mlst_data)

mlst_results_list.append({
    'Variant': 'All',
    'r_s_MLST': round(overall_spearman, 3),
    'p-value_MLST': overall_spearman_p,
    'R-squared_MLST': round(overall_r_squared, 3)
})

# ----------------------------------------------------
# STORE NUMERIC MLST TABLE
# ----------------------------------------------------
table2_mlst_df = pd.DataFrame(mlst_results_list)

# Force final row order
variant_order = ['All', 'Alpha', 'Epsilon', 'Beta', 'Gamma', 'Delta', 'Omicron']

table2_mlst_df['Variant'] = pd.Categorical(
    table2_mlst_df['Variant'],
    categories=variant_order,
    ordered=True
)

table2_mlst_df = table2_mlst_df.sort_values('Variant').reset_index(drop=True)

# ----------------------------------------------------
# CREATE MANUSCRIPT-DISPLAY TABLE WITH FORMATTED P-VALUES
# ----------------------------------------------------
def format_p_value(p):
    if pd.isna(p):
        return ""
    if p == 0:
        return "<1 × 10^-300"
    exponent = int(np.floor(np.log10(abs(p))))
    coefficient = p / (10 ** exponent)
    return f"{coefficient:.2f} × 10^{exponent}"

table2_mlst_display_df = table2_mlst_df.copy()
table2_mlst_display_df['p-value_MLST'] = table2_mlst_display_df['p-value_MLST'].apply(format_p_value)

table2_mlst_display_df
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from scipy.stats import spearmanr
from matplotlib.gridspec import GridSpec
from matplotlib.ticker import MaxNLocator
from matplotlib import font_manager

# -----------------------------
# Font handling
# -----------------------------
available_fonts = [f.name for f in font_manager.fontManager.ttflist]

if 'Arial' in available_fonts:
    chosen_font = 'Arial'
else:
    chosen_font = 'Times New Roman'   # good fallback on Linux/Colab

plt.rcParams.update({
    'font.family': chosen_font,
    'font.size': 9,
    'axes.labelsize': 9,
    'axes.titlesize': 9,
    'xtick.labelsize': 9,
    'ytick.labelsize': 9,
    'legend.fontsize': 9
})

# -----------------------------
# Data columns
# -----------------------------
cli_columns = ['delta_21A_percentage', 'delta_21A_level_1_percentage', 'delta_21I_percentage', 'delta_21I_level_1_percentage', 'delta_21I_level_2_percentage', 'delta_21J_percentage', 'delta_21J_level_1_percentage', 'delta_21J_level_2_percentage']
ww_columns = ['Delta_21A%', 'Delta_21A_level_1%', 'Delta_21I%', 'Delta_21I_level_1%', 'Delta_21I_level_2%', 'Delta_21J%', 'Delta_21J_level_1%', 'Delta_21J_level_2%']
clade_name = ['Delta_21A', 'Delta_21A_level_1', 'Delta_21I', 'Delta_21I_level_1', 'Delta_21I_level_2', 'Delta_21J', 'Delta_21J_level_1', 'Delta_21J_level_2']
colors = ['tomato', 'orange', 'blue', 'blueviolet', 'deepskyblue', 'olive', 'lightseagreen', 'lime']

results_list = []
all_r_squared = []

# -----------------------------
# Figure and layout
# -----------------------------
fig = plt.figure(figsize=(7.5, 8), dpi=300)
outer = GridSpec(3, 1, figure=fig, hspace=0.55)

# Row 1: 2 panels
gs_row1 = outer[0].subgridspec(1, 2, wspace=0.35)
# Row 2: 3 panels
gs_row2 = outer[1].subgridspec(1, 3, wspace=0.35)
# Row 3: 3 panels
gs_row3 = outer[2].subgridspec(1, 3, wspace=0.35)

axes = [
    fig.add_subplot(gs_row1[0, 0]),  # Delta_21A
    fig.add_subplot(gs_row1[0, 1]),  # Delta_21A_level_1
    fig.add_subplot(gs_row2[0, 0]),  # Delta_21I
    fig.add_subplot(gs_row2[0, 1]),  # Delta_21I_level_1
    fig.add_subplot(gs_row2[0, 2]),  # Delta_21I_level_2
    fig.add_subplot(gs_row3[0, 0]),  # Delta_21J
    fig.add_subplot(gs_row3[0, 1]),  # Delta_21J_level_1
    fig.add_subplot(gs_row3[0, 2])   # Delta_21J_level_2
]

panel_labels = list("ABCDEFGH")

# -----------------------------
# Plot loop
# -----------------------------
for i, ((cli_col, ww_col), color, name) in enumerate(zip(zip(cli_columns, ww_columns), colors, clade_name)):
    ax = axes[i]

    X = combined_study_cleandf[[cli_col]].dropna()
    y = combined_study_cleandf[[ww_col]].loc[X.index]

    valid_indices = X.join(y).dropna().index
    X = X.loc[valid_indices]
    y = y.loc[valid_indices]

    x_vals = X[cli_col].to_numpy()
    y_vals = y[ww_col].to_numpy()

    # Spearman correlation
    corr, p = spearmanr(x_vals, y_vals)

    # Linear regression
    model = LinearRegression().fit(x_vals.reshape(-1, 1), y_vals.reshape(-1, 1))
    y_pred = model.predict(x_vals.reshape(-1, 1)).flatten()
    r_squared = r2_score(y_vals, y_pred)

    # Save results
    results_list.append({'Variant': name, 'r_s': corr, 'R^2': r_squared, 'p-value': p})
    all_r_squared.append(r_squared)

    # Sort for regression line
    sort_idx = np.argsort(x_vals)
    x_sorted = x_vals[sort_idx]
    y_pred_sorted = y_pred[sort_idx]

    # Scatter + line
    ax.scatter(x_vals, y_vals, color=color, s=22)
    ax.plot(x_sorted, y_pred_sorted, color='black', linewidth=1.2)

    # Title
    ax.set_title(
        f'{name}\nR² = {r_squared:.3f}, r$_s$ = {corr:.3f}',
        fontname=chosen_font,
        fontsize=9,
        fontweight='bold',
        pad=4
    )

    # Panel letter
    ax.text(
        0.02, 0.98, panel_labels[i],
        transform=ax.transAxes,
        ha='left',
        va='top',
        fontsize=10,
        fontweight='bold',
        fontname=chosen_font
    )

    # Tick density
    ax.xaxis.set_major_locator(MaxNLocator(nbins=4))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=5))

    # Special fix for Delta_21I_level_2
    if name == 'Delta_21I_level_2':
        ax.xaxis.set_major_locator(MaxNLocator(nbins=3))
        ax.tick_params(axis='x', rotation=25)

    # Tick appearance
    ax.tick_params(axis='both', which='major', labelsize=9, width=1, length=4)

    for tick in ax.get_xticklabels():
        tick.set_fontname(chosen_font)
        tick.set_fontweight('bold')

    for tick in ax.get_yticklabels():
        tick.set_fontname(chosen_font)
        tick.set_fontweight('bold')

    # Offset text
    ax.xaxis.get_offset_text().set_fontsize(8)
    ax.yaxis.get_offset_text().set_fontsize(8)
    ax.xaxis.get_offset_text().set_fontweight('bold')
    ax.yaxis.get_offset_text().set_fontweight('bold')

    # Axis spine thickness
    for spine in ax.spines.values():
        spine.set_linewidth(1)

# -----------------------------
# Shared labels
# -----------------------------
fig.supxlabel(
    'Relative abundances of Delta sub variants in clinical samples',
    fontsize=9,
    fontweight='bold',
    fontname=chosen_font,
    y=0.03
)

fig.supylabel(
    'Relative abundances of Delta sub variants in wastewater samples',
    fontsize=9,
    fontweight='bold',
    fontname=chosen_font,
    x=0.02
)

# -----------------------------
# Overall statistics
# -----------------------------
cli_flat = combined_study_cleandf[cli_columns].values.flatten()
ww_flat = combined_study_cleandf[ww_columns].values.flatten()

valid_mask = ~np.isnan(cli_flat) & ~np.isnan(ww_flat)
overall_corr, overall_p = spearmanr(cli_flat[valid_mask], ww_flat[valid_mask])
overall_r_squared = np.mean(all_r_squared)

results_list.append({
    'Variant': 'Overall',
    'r_s': overall_corr,
    'R^2': overall_r_squared,
    'p-value': overall_p
})

results = pd.DataFrame(results_list)

# -----------------------------
# Save + show
# -----------------------------
plt.savefig('/content/drive/MyDrive/Freyja_Manuscript/Master_File/Delta_subvariants/Final_delta_clade_ay_level_variants/Output/Delta_multipanel_3row_PLOS.png', dpi=300, bbox_inches='tight')
plt.savefig('/content/drive/MyDrive/Freyja_Manuscript/Master_File/Delta_subvariants/Final_delta_clade_ay_level_variants/Output/Delta_multipanel_3row_PLOS.tiff', dpi=300, bbox_inches='tight')
plt.show()

results
```
